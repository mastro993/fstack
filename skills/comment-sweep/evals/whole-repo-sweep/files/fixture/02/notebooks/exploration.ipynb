# Exploring July readings

Station coverage, and the °F/°C duplicates SkyFeed sends for some stations.

```python
%matplotlib inline
import pandas as pd
from sqlalchemy import create_engine

from wxdata.config import load_settings
```

```python
# Load the data
settings = load_settings()
engine = create_engine(settings.database_url)
df = pd.read_sql("SELECT * FROM readings WHERE observed_at >= '2024-07-01'", engine)
# df.describe()
df.head()
```

## Coverage

A station should deliver 288 readings a day, one every five minutes.

```python
per_day = df.groupby(["station_id", df["observed_at"].dt.date]).size()
per_day.unstack(0).plot(figsize=(12, 4), legend=False)
```

## Duplicates

```python
# Count rows per station and timestamp
dupes = df.groupby(["station_id", "observed_at"]).size()
dupes[dupes > 1].sort_values(ascending=False).head(20)
```
